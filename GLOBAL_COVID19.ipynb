### Adding required columns for data analysis

```python
# adding 'doses per person' column
country_summary["doses_per_person"] = country_summary["total_vaccinations"]/country_summary["population"]
```

```python
country_summary.describe()
```

```python
# Classify vaccination and fatality outcomes using a custom function

def vaccination_fatality(row):
# 75% 'doses-per-person' (>2.16) & 25% 'case-fatality-rate'(<0.24)
    if row["doses_per_person"] > 2.16 and row["case_fatality_rate"] <0.24:
        return "High Vaccination & Low Fatality"
        
# 25% 'doses-per-person'(<0.70) & 75% 'case-fatality-rate'(>1.61)
    if row["doses_per_person"] <0.70 and row["case_fatality_rate"] >1.61:
        return "Low Vaccination & High Fatality"

    return "Mid Range"
```

```python
# apply the custom funcction and add the 'vaccination_outcome' column

country_summary["vaccination_outcome"] = country_summary.apply(vaccination_fatality, axis="columns")
country_summary.head()
```

```python
unprotected = country_summary[country_summary["vaccination_outcome"] == "Low Vaccination & High Fatality"]
```

```python
well_protected = country_summary[country_summary["vaccination_outcome"] == "High Vaccination & Low Fatality"]
```

#### 8. Which locations recorded high vaccination with low fatality?

```python
# Findings : 'Yemen', 'Sudan', 'Syria', 'Somalia' and 'Bosnia  and Herzegovina' recorded low vaccination and high fatality.
```

#### 9.Which locations recorded low vaccination with high fatality?

```python
well_protected.sort_values("case_fatality_rate", ascending=False).head(5)
```

```python
# Findings : 'Japan', 'Bahrain', 'UAE', 'Australia' and 'Guernsey' recorded high vaccination and low fatality.
```
